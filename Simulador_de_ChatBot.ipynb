```python
# CONFIGURAÇÃO DO CONECTOR E ROTEADOR
from langchain.chains import ConversationalRetrievalChain

DEFAULT_FALLBACK = (
    "Desculpe, a informação exata sobre este tópico não está detalhada "
    "nas Políticas de Uso da iAutos que eu tenho acesso."
)

# Criação da chain principal
qa_chain = ConversationalRetrievalChain.from_llm(
    llm=llm,
    retriever=vectordb.as_retriever(search_kwargs={"k": 3}),
    return_source_documents=False
)

# --- Função de roteamento inteligente ---
def handle_query(query):
    """
    Direciona a pergunta para o lugar certo.
    """
    q_lower = query.lower()

    # Detecta temas fora do escopo
    if any(palavra in q_lower for palavra in ["preço", "valor", "quanto custa", "gol", "caro", "barato"]):
        return "A iAutos não define preços de veículos. Os valores variam conforme os anúncios de vendedores. Consulte a barra de busca da plataforma para ver os preços atualizados."

    # Se for dúvida sobre políticas, envia ao modelo
    result = qa_chain.invoke({"question": query, "chat_history": []})
    return result["answer"]

print("✅✅✅ CONECTOR CRIADO E ROTEADOR ATIVO!")
```

```python
# BATERIA DE TESTES AUTOMÁTICOS
def run_pipeline_tests(verbose=True):
    tests = [
        {"query": "O que é a iAutos?", "expect_contains": ["marketplace", "classificados", "iAutos"]},
        {"query": "Quais são os tipos de veículos permitidos?", "expect_contains": ["automóveis", "motos", "utilitários"]},
        {"query": "Qual o preço de um Gol?", "expect_contains": ["não define preços", "anúncios", "valores variam"]},  # <- Novo teste
        {"query": "Como eu entro em contato com o vendedor?", "expect_contains": ["contato", "vendedor", "mensagem"]},
        {"query": "O que acontece se eu fraudar?", "expect_contains": ["banido", "banimento", "fraudar"]}
    ]

    passed, failed = 0, 0
    details = []

    print("\n--- 🧪 Iniciando Bateria de Testes Automáticos ---\n")

    for t in tests:
        q = t["query"]
        try:
            ans = handle_query(q)
            ans_lower = ans.lower()
            expected = any(word in ans_lower for word in t["expect_contains"])

            if expected:
                passed += 1
                status = "✅ PASSOU"
            else:
                failed += 1
                status = "❌ FALHOU"

            print(f"[{status}] Pergunta: {q}\n--> Resposta: {ans[:100]}...\n")
            details.append({"query": q, "status": status, "answer_sample": ans[:100]})

        except Exception as e:
            failed += 1
            print(f"[ERRO] Pergunta: {q} -> {e}\n")
            details.append({"query": q, "status": "ERRO", "error": str(e)})

    print(f"--- 🏁 Resumo dos Testes ---")
    print(f"✅ Passou: {passed} | ❌ Falhou: {failed}\n")
    return passed, failed, details

# --- Executa ---
run_pipeline_tests()




```

```python
## ==============================================================================
## SEÇÃO 2: PROCESSO FINAL (TUDO EM UMA CÉLULA)
## ==============================================================================

## --- 1. Importações necessárias ---
import os
import gdown
from google.colab import userdata
from langchain_openai import AzureChatOpenAI, AzureOpenAIEmbeddings
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.vectorstores import Chroma
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.chains import ConversationalRetrievalChain
from langchain.memory import ConversationBufferMemory
from langchain.prompts.chat import (
    ChatPromptTemplate,
    HumanMessagePromptTemplate,
    SystemMessagePromptTemplate,
    MessagesPlaceholder
)
import warnings

# Ignorar avisos de depreciação do LangChain (opcional, mas limpa o output)
warnings.filterwarnings("ignore")

## --- 2. Definição de Todas as Funções ---

def baixar_documento_contexto():
    """Baixa o arquivo PDF do Google Drive."""
    file_id = "1-ZpUOl8OA4lxa8CJ6auT42hSxaF3jclk"
    local_filename = "iAutos_Politicas_e_Uso.pdf"
    print(f"\n📥 Baixando arquivo de contexto: {local_filename}")
    try:
        gdown.download(f"https://drive.google.com/uc?id={file_id}", local_filename, quiet=True)
        print("✅ Arquivo baixado com sucesso.")
        return local_filename
    except Exception as e:
        print(f"❌ Erro ao baixar o arquivo: {e}")
        return None

def preparar_documentos(pdf_path):
    """Carrega o PDF e divide em chunks."""
    if not pdf_path:
        return []
    print("\n📄 Carregando e fatiando o documento...")
    loader = PyPDFLoader(pdf_path)
    documents_pdf = loader.load()
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200,
        length_function=len,
    )
    chunks = text_splitter.split_documents(documents_pdf)
    print(f"✅ Documento dividido em {len(chunks)} fragmentos.")
    return chunks

def criar_vectordb(chunks):
    """Cria o ChromaDB usando Azure Embeddings (com Secrets)."""
    print("\n🧠 Criando o VectorDB (ChromaDB) com Azure Embeddings...")
    os.environ["AZURE_OPENAI_ENDPOINT"] = userdata.get('AZURE_OPENAI_ENDPOINT')
    os.environ["AZURE_OPENAI_API_KEY"] = userdata.get('AZURE_OPENAI_API_KEY')

    embeddings = AzureOpenAIEmbeddings(
        azure_deployment=userdata.get('EMB_DEPLOYMENT_MODEL'),
        openai_api_version=userdata.get('EMB_API_VERSION')
    )
    vectordb = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        collection_name="iAutos_politicas"
    )
    print("✅ ChromaDB criado e pronto para recuperação (Retriever).")
    return vectordb

def construir_chatbot_chain(vectordb):
    """Constrói o ChatBot com LLM, Memória e o Prompt corrigido."""
    print("\n🤖 Construindo o Chatbot (ConversationalRetrievalChain)...")

    os.environ["AZURE_OPENAI_ENDPOINT"] = userdata.get('AZURE_OPENAI_ENDPOINT')
    os.environ["AZURE_OPENAI_API_KEY"] = userdata.get('AZURE_OPENAI_API_KEY')

    llm = AzureChatOpenAI(
        azure_deployment=userdata.get('LLM_DEPLOYMENT_MODEL'),
        openai_api_version=userdata.get('LLM_API_VERSION'),
        temperature=0.3
    )

    memory = ConversationBufferMemory(
        memory_key="chat_history",
        return_messages=True,
        output_key="answer"
    )

    # --- PROMPT COMPATÍVEL COM STUFFDOCUMENTSCHAIN ---
    from langchain.prompts import PromptTemplate

    SYSTEM_TEMPLATE = """
Você é o ChatBot de Atendimento da iAutos, um marketplace de classificados de veículos.
Sua missão é responder perguntas dos clientes de forma precisa, educada e SOMENTE com base no contexto
fornecido abaixo, que contém as políticas de uso e regras de publicação de veículos da iAutos.

Diretrizes de Resposta:
- Responda SOMENTE com informações encontradas no contexto recuperado.
- Se o contexto **não contiver** a resposta, diga EXATAMENTE:
  "Desculpe, a informação exata sobre este tópico não está detalhada nas Políticas de Uso da iAutos que eu tenho acesso."
- Nunca invente ou faça suposições, mesmo que a pergunta pareça simples.
- Mantenha o tom de voz profissional, educado e prestativo.

Contexto:
{context}

Pergunta:
{question}
"""

    qa_prompt = PromptTemplate(
        template=SYSTEM_TEMPLATE,
        input_variables=["context", "question"]
    )

    qa_chain = ConversationalRetrievalChain.from_llm(
        llm=llm,
        retriever=vectordb.as_retriever(search_kwargs={"k": 3}),
        memory=memory,
        combine_docs_chain_kwargs={"prompt": qa_prompt},
        return_source_documents=True,
        output_key="answer"
    )

    print("✅ ChatBot pronto!")
    return qa_chain

def simular_conversa(qa_chain):
    """
    Loop de conversação para interagir com o ChatBot.
    AGORA COM O ROTEADOR INTELIGENTE.
    """
    print("\n\n--- 🤖 SIMULADOR DE CHATBOT iAutos (Versão Robusta) ---")
    print("Digite 'sair' a qualquer momento para encerrar.\n")

    # A memória já está dentro da qa_chain, então não precisamos
    # gerenciar o histórico aqui fora.

    while True:
        try:
            query = input("[VOCÊ]: ")
            if query.lower() in ["sair", "exit", "tchau", "fim", "encerrar"]:
                print("\n[CHATBOT]: Até logo! Obrigado por usar a iAutos.")
                break

            # --- ROTEADOR INTELIGENTE (A lógica do teste) ---
            q_lower = query.lower()
            if any(palavra in q_lower for palavra in ["preço", "valor", "quanto custa", "gol", "caro", "barato"]):
                # AÇÃO A: Responde com a regra de negócio
                print("\n[CHATBOT]: A iAutos não define preços de veículos. Os valores variam conforme os anúncios de vendedores. Consulte a barra de busca da plataforma para ver os preços atualizados.")

            else:
                # AÇÃO B: Usa o RAG para perguntas de política
                result = qa_chain.invoke({"question": query})
                print("\n[CHATBOT]:", result['answer'])
            # --- FIM DO ROTEADOR ---

        except Exception as e:
            print(f"\n[CHATBOT]: Ops, houve um erro inesperado. Erro: {e}")
            break # Encerra o loop em caso de erro# Encerra o loop em caso de erro

def rodar_pipeline_completo():
    """Executa todas as etapas do processo final."""

    # A instalação foi movida para a célula anterior

    pdf_path = baixar_documento_contexto()
    if pdf_path:
        chunks = preparar_documentos(pdf_path)
        if chunks:
            vectordb = criar_vectordb(chunks)
            qa_chain = construir_chatbot_chain(vectordb)
            simular_conversa(qa_chain)
        else:
            print("❌ Erro: Nenhum fragmento (chunk) foi criado do PDF.")
    else:
        print("❌ Erro: O download do PDF falhou. O pipeline não pode continuar.")

## --- 3. Execução Imediata ---
# (Roda o pipeline assim que esta célula é executada)
rodar_pipeline_completo()
```
